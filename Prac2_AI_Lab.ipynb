## Prac 2: BFS,DFS and uniform cost search on graph or maze prob : compare search cost and path quality

```python
print("Hello, I'll take you to the GOAL")
mapRooms = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F'],
    'D': [],
    'E': ['F'],
    'F': ['E']
}

startRoom = 'A'
goalRoom = 'F'

print("Starting Room:", startRoom)
print("Goal Room:", goalRoom)
print("Map of rooms:" ,mapRooms)
print("            A          ")
print("           / \         ")
print("          B   C        ")
print("         / \   \       ")
print("        D   E   F      ")
```

```python
def bfs(start, goal):
  queue = [start]
  visited =[]
  order =[]
  while queue:
    current = queue.pop(0)
    if current in visited:
      continue
    visited.append(current)
    order.append(current)

    if current == goal:
      return order

    for neighbor in mapRooms[current]:
      queue.append(neighbor)

  return order

order = bfs("A","F")
print(order)
print(len(order))
```

```python
def dfs(start, goal):
  stack = [start]
  visited =[]
  order =[]
  while stack:
    current = stack.pop()
    visited.append(current)
    order.append(current)
    if current == goal:
      return order

    for neighbor in mapRooms[current]:
        stack.append(neighbor)

  return order

order = dfs("A","F")
print(order)
print(len(order))
print(order)
```

```python
doors = {
    ('A', 'B'): 1,
    ('A', 'C'): 2,
    ('B', 'D'): 3,
    ('B', 'E'): 4,
    ('C', 'F'): 5
}

def path_cost(path):
    cost = 0
    for i in range(len(path) - 1):
        current_room = path[i],path[i+1]
        cost = cost+doors[current_room]
    return cost

path1 = ['A', 'B', 'D']
path2 = ['A', 'C', 'F']

cost1 = path_cost(path1)
cost2 = path_cost(path2)

print("Path 1:", path1)
print("Cost of Path 1:", cost1)

print("Path 2:", path2)
print("Cost of Path 2:", cost2)
```

```python

if cost1<cost2:
  best_cost = cost1
  best_path = path1
  print("Path 1 is better than Path 2")
else:
  print("Path 2 is better than Path 1")

print(best_cost)
```
